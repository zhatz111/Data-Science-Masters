First we start by exploring the trout data's distribution to understand how the data was collected and what shape we can expect from these observations.

```python
pd.Series(trout).describe()
```

```python
plt.figure(figsize=(8,4))
plt.hist(trout, bins=10)
plt.xlabel("trout")
plt.ylabel("count")
plt.title("Distribution of trout")
plt.show()
```

Based on the shape of the data being normally distributed, a 95% confidence interval can be constructed using the standard error and the z-score. This is the appropriate approach compared to a bootstrap approach because it is simpler and we can show the data is distributed normally.

```python
mean, std = np.mean(trout), np.std(trout)
sem = std / np.sqrt(len(trout))
ci_95 = 1.96 * sem
print(f"Mean Model with 95% Confidence Interval: {mean:.2f} ± {ci_95:.2f}")
print(f"Bounds: [{mean - ci_95:.2f}, {mean + ci_95:.2f}]")
```

Q3 - The trout data was most likely generated by randomly sampling a normal distribution.

Q4 - The three questions I will answer using a distributional model are the following:
- What is the probability of trout > 500
- What is the value at the 90th percentile
- Give a 95% prediction interval for a single new fish

```python
# Distributional Model of trout data
from scipy import stats
mu, sigma = np.mean(trout), np.std(trout, ddof=1)
lo, hi = stats.norm.ppf([0.025, 0.975], loc=mu, scale=sigma)
print(f"Distributional Model with 95% Prediction Interval: {mu:.2f} ± {sigma:.2f}")
```
